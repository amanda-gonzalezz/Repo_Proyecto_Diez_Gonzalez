Este notebook carga la base de datos ya limpia (`base_gasto_core_2024_limpia.csv`) y muestra sus primeras filas, tal como pide el enunciado. El proceso completo de limpieza (filtrado, agregación y cruce) está documentado paso a paso en el `README.md` de esta carpeta.

```python
import pandas as pd

# Cargar la base de datos limpia
df = pd.read_csv('../base_gasto_core_2024_limpia.csv')

# Ver dimensiones y primeras filas
print('Dimensiones de la base:', df.shape)
df.head()
```

```python
# Chequeo rápido de calidad: nulos y duplicados
print('Valores nulos por columna:')
print(df.isnull().sum())
print()
print('Duplicados de id_candidato:', df['id_candidato'].duplicated().sum())
```

```python
# Ejemplo de tabla dinámica: gasto promedio por pacto electoral (top 5)
df.groupby('pacto')['gasto_total_declarado'].mean().sort_values(ascending=False).head()
```

```python
# Ejemplo de tabla dinámica: cuántos candidatos declararon gasto y cuántos no
df['declara_gasto'].value_counts()
```

```python
# Ejemplo de tabla dinámica: relación entre estado de auditoría y si declaró gasto
pd.crosstab(df['estado_cuenta_auditoria'], df['declara_gasto'])
```

```python
# Ejemplo de tabla dinámica: gasto promedio por circunscripción (top 5)
df.groupby('circunscripcion')['gasto_total_declarado'].mean().sort_values(ascending=False).head()
```

```python
# Chequeo de outliers con el método IQR (visto en clase)
q1 = df['gasto_total_declarado'].quantile(0.25)
q3 = df['gasto_total_declarado'].quantile(0.75)
iqr = q3 - q1
limite_superior = q3 + 1.5 * iqr
print(f'Límite superior (Q3 + 1.5*IQR): {limite_superior:,.0f}')
print(df['outlier_gasto_iqr'].value_counts())

# Nota: estos 163 casos NO se eliminaron de la base, ya que corresponden a
# candidatos reales con gasto alto (no a errores de datos), relevantes para
# responder la pregunta de investigación sobre quién gastó más.
```
